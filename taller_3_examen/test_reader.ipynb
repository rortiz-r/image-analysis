```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
import os
```

```python
im = cv.cvtColor(cv.imread("../AI-HR01/AI-HR01.jpg"), cv.COLOR_BGR2RGB)
plt.imshow(im)
plt.xticks([])
plt.yticks([])
```

```python
im_gray = cv.cvtColor(im, cv.COLOR_BGR2GRAY)

## Aplicamos un filtro gausiano para suavizar

im_gray = cv.GaussianBlur(im_gray, (5,5),0)

plt.imshow(im_gray, cmap='gray')
```

```python
x = 450
y = 600
w = x+1000
h = y+2000

im_gray = cv.rectangle(im_gray, (x, y), (w, h), (0, 0, 0), 4)
```

```python


#Cropped image

cropped_image = im_gray[y:h, x:w]


plt.imshow(cropped_image, cmap="gray")
```

```python
### Detectar circulos


_, thresh1 = cv.threshold(cropped_image, 200, 255, cv.THRESH_BINARY_INV + cv.THRESH_OTSU)


plt.imshow(thresh1, cmap="gray")
```

### Detectar contornos

```python
thres1_copy = thresh1.copy()

contours,_ = cv.findContours(thres1_copy, cv.RETR_EXTERNAL, cv.CHAIN_APPROX_SIMPLE)

result = np.zeros(thres1_copy.shape)

cv.drawContours(result, contours[0], -1, (255,255,255), 3)



plt.imshow(result, cmap="gray")
```
