```python
from fpdf import FPDF

class PDF(FPDF):
    def header(self):
        self.set_font('Arial', 'B', 15)
        self.cell(0, 10, 'Guía de Comandos Cisco (Cheat Sheet)', 0, 1, 'C')
        self.ln(10)

    def chapter_title(self, label):
        self.set_font('Arial', 'B', 12)
        self.set_fill_color(200, 220, 255)
        self.cell(0, 6, f'{label}', 0, 1, 'L', 1)
        self.ln(4)

    def chapter_body(self, body):
        self.set_font('Arial', '', 11)
        self.multi_cell(0, 5, body)
        self.ln()

    def code_block(self, code):
        self.set_font('Courier', '', 10)
        self.set_fill_color(240, 240, 240) # Gris claro
        self.multi_cell(0, 5, code, 1, 'L', True)
        self.ln(10)

pdf = PDF()
pdf.add_page()

# 1. OSPF
pdf.chapter_title('OSPF (Open Shortest Path First)')
pdf.chapter_body('Qué debo hacer:\n- Habilitar el proceso OSPF (ID proceso).\n- Anunciar redes conectadas.\n- Usar Wildcard Mask.\n- Asignar área (normalmente 0).')
pdf.code_block("""router ospf 1
 network 192.168.1.0 0.0.0.255 area 0
 network 10.0.0.0 0.0.0.3 area 0""")

# 2. Router on a Stick
pdf.chapter_title('Enrutamiento entre VLANs (Router on a Stick)')
pdf.chapter_body('Qué debo hacer:\n- Router: Crear subinterfaces, encapsulación 802.1Q, asignar IP.\n- Switch: Crear VLANs y poner el puerto al router en modo Trunk.')
pdf.code_block("""! --- ROUTER ---
interface g0/0.10
 encapsulation dot1Q 10
 ip address 192.168.10.1 255.255.255.0
interface g0/0.20
 encapsulation dot1Q 20
 ip address 192.168.20.1 255.255.255.0

! --- SWITCH ---
vlan 10 name Usuarios
vlan 20 name Servidores
interface g0/1
 switchport mode trunk
 switchport trunk allowed vlan 10,20""")

# 3. RIP v2
pdf.chapter_title('RIP v2 (Routing Information Protocol)')
pdf.chapter_body('Qué debo hacer:\n- Habilitar RIP versión 2.\n- Desactivar auto-summary.\n- Anunciar redes (sin máscara ni wildcard).')
pdf.code_block("""router rip
 version 2
 no auto-summary
 network 192.168.10.0
 network 10.0.0.0""")

# 4. EIGRP
pdf.chapter_title('EIGRP (Enhanced Interior Gateway Routing Protocol)')
pdf.chapter_body('Qué debo hacer:\n- Habilitar EIGRP con AS (ej. 100).\n- Anunciar redes conectadas.\n- Usar Wildcard Mask.')
pdf.code_block("""router eigrp 100
 network 192.168.10.0 0.0.0.255
 network 10.1.0.0 0.0.255.255""")

pdf.output('Guia_Comandos_Cisco.pdf', 'F')
print("PDF Generado exitosamente.")
```
