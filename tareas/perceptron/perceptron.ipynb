```python
## Inicializamos los pesos

def initialize_parameters(n_h = 1, n_x = 2, n_y = 1):
    scale_factor = 1
    w_1 = np.random.randn(n_h, n_x) * scale_factor
    b1 = np.zeros((n_h, 1))
    w_2 = np.random.randn(n_y, n_h) * scale_factor
    b2 = np.zeros((n_y, 1))


    parameters = {"W1": w_1,
                  "b1": b1,
                  "W2": w_2,
                  "b2": b2}
    
    return w_1, b1, w_2, b2
```

# Definimos nuestra función de activación

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))

def relu(x):
    return np.maximum(0, x)
```

### Forward propagation

```python
def forward(X, w_1, b1, w_2, b2):
    Z1 = np.dot(w_1, X) + b1 
    A1 = np.tanh(Z1) 
    Z2 = np.dot(w_2, A1) + b2
    A2 = sigmoid(Z2)
    return Z1, A1, Z2, A2

```

```python
def cost(A2, Y):##
    m = Y.shape[1]
    logprobs =   np.multiply(np.log(A2),Y)+np.multiply((1-Y),np.log(1-A2)) 
    cost = - np.sum(logprobs)/m
    cost = float(np.squeeze(cost))
    return cost
```

```python
def backward_propagation(w1, w2, b1, b2, A1,A2, X,Y):
    m = X.shape[1]
    dZ2 = A2 - Y
    dW2 = (1/m) * np.dot(dZ2, A1.T)
    db2 = (1/m) * np.sum(dZ2, axis=1, keepdims=True)
    dZ1 = np.multiply(np.dot(w2.T, dZ2), 1 - np.power(A1, 2))
    dW1 = (1/m) * np.dot(dZ1, X.T)
    db1 = (1/m) * np.sum(dZ1, axis=1, keepdims=True)

    grads = {
        "dw1": dW1, "dw2": dW2, "db1": db1, "db2": db2
    }
    return grads
```

```python
def update_parameters(w1, w2, b1, b2, grads, learning_rate):

    dW1 = grads.get("dw1") 
    dW2 = grads.get("dw2") 
    db1 = grads.get("db1") 
    db2 = grads.get("db2") 

    w1 -= learning_rate*dW1
    w2 -= learning_rate*dW2
    db1 -= learning_rate*db1
    db2 -= learning_rate*db2

    return {"w1":w1, "w2":w2, "b1":b1, "b2":b2}
```

```python
def neural_network(X,Y, n_h, learning_rate, num_iterations=500, print_cost=False):
    n_x = x.shape[0]
    n_y = y.shape[0]
    w1,b1,w2,b2 = initialize_parameters(n_h, n_x, n_y)
    for i in range(0,num_iterations):
        Z1, A1, Z2, A2, = forward(X, w1,b1, w2,b2)
        computed_cost = cost(A2, Y)
        grads = backward_propagation(w1,w2,b1,b2,A1,A2,X,Y)
        updated_parameters = update_parameters(w1,w2,b1,b2, grads, learning_rate)
        if print_cost and i % 1000 == 0:
            print(f"Cost on iteration {i}: {computed_cost} {np.mean(w1), np.mean(w2)}")

    

    return updated_parameters
    

```

```python
parameters = neural_network(x, y, 2 ,0.1,num_iterations=20000,print_cost=True)
```
